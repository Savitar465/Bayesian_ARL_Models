## 1. Descripción

Una **Bayesian Network** es un modelo gráfico probabilístico que representa relaciones de dependencia condicional entre variables mediante un grafo dirigido acíclico. Su utilidad principal es modelar incertidumbre, calcular probabilidades posteriores y realizar inferencia con evidencia parcial.

A partir de un conjunto de variables observables, la red describe cómo se factoriza la distribución conjunta en un producto de distribuciones condicionales más pequeñas.

### Casos de uso típicos
- Diagnóstico y pronóstico.
- Toma de decisiones bajo incertidumbre.
- Análisis de riesgo.
- Sistemas expertos.
- Modelado causal aproximado.

> **Nota — Bayesian Network vs BBN:** una *Bayesian Network* y una *Bayesian Belief Network* son el **mismo modelo** (grafo dirigido acíclico + tablas de probabilidad condicional). Aquí se enfatiza la **factorización de la distribución conjunta** $P(X_1,\dots,X_n)=\prod_i P(X_i \mid \text{Pa}(X_i))$ y el razonamiento **causal**; el notebook de *BBN* aborda el mismo formalismo desde la óptica de **diagnóstico/creencias**.

## 4. Algoritmo de Entrenamiento

El entrenamiento de una red bayesiana incluye definir la estructura del grafo y estimar las distribuciones condicionales de cada nodo.

### Pseudocódigo
```text
Entrada: datos D, estructura G
Salida: red bayesiana

1. Definir la estructura del grafo o aprenderla a partir de datos.
2. Para cada variable X_i:
   a. Determinar sus padres Pa(X_i).
   b. Estimar P(X_i | Pa(X_i)).
3. Almacenar las tablas de probabilidad condicional.
4. Para inferencia, incorporar evidencia y propagar probabilidades posteriores.
```

### Componentes principales
- **Nodos:** variables aleatorias.
- **Arcos dirigidos:** dependencias condicionales.
- **CPTs:** tablas de probabilidad condicional.
- **Inferencia:** cálculo de probabilidades posteriores.

### Métricas clave
- **Probabilidad posterior:** $P(H \mid E)$
- **Log-loss:** calidad de las probabilidades
- **Exactitud:** aciertos sobre una variable objetivo
- **Matriz de confusión:** resumen de clasificación inferida

## 5. Supuestos y Restricciones

- **Grafo acíclico dirigido:** la red no puede contener ciclos.
- **Relaciones condicionales bien especificadas:** cada nodo depende de sus padres.
- **Datos suficientes:** las CPT requieren suficiente soporte para ser fiables.
- **Evidencia parcial:** el modelo debe poder inferir con variables observadas e invisibles.
- **Complejidad de inferencia:** algunas consultas exactas son costosas.
- **Manejo de datos faltantes:** debe resolverse con imputación o métodos de inferencia.
- **Calidad de estructura:** una estructura mal definida afecta la interpretación y el desempeño.

## 6. Tests / Métricas de validación

Las redes bayesianas se evalúan mediante métricas probabilísticas y de clasificación:

- **Exactitud** — porcentaje de aciertos.
- **Precision / Recall / F1** — desempeño en la variable objetivo.
- **Log-loss** — calidad de la probabilidad predicha.
- **Matriz de confusión** — errores y aciertos por clase.
- **Calibración** — coherencia entre probabilidades y frecuencias observadas.

## 7. Implementación práctica

Usaremos una red de ejemplo simple con inferencia manual para ilustrar el razonamiento bayesiano.

### 7.1 Instalación e imports

````python
import pandas as pd

print('Librerías cargadas correctamente')
````

### 7.2 Red de ejemplo

Modelaremos un escenario simple de evaluación de riesgo crediticio con evidencia observada.

````python
prior_riesgo = {'alto': 0.25, 'bajo': 0.75}
p_ingresos = {'alto': 0.3, 'bajo': 0.8}
p_deuda = {'alto': 0.7, 'bajo': 0.2}
p_historial = {'alto': 0.6, 'bajo': 0.1}

evidencia = {'ingresos': 'alto', 'deuda': 'alto', 'historial': 'alto'}
evidencia
````

### 7.3 Inferencia bayesiana

````python
def inferir_riesgo(evidencia):
    score_alto = prior_riesgo['alto']
    score_bajo = prior_riesgo['bajo']

    score_alto *= p_ingresos['alto'] if evidencia['ingresos'] == 'alto' else (1 - p_ingresos['alto'])
    score_bajo *= p_ingresos['bajo'] if evidencia['ingresos'] == 'alto' else (1 - p_ingresos['bajo'])

    score_alto *= p_deuda['alto'] if evidencia['deuda'] == 'alto' else (1 - p_deuda['alto'])
    score_bajo *= p_deuda['bajo'] if evidencia['deuda'] == 'alto' else (1 - p_deuda['bajo'])

    score_alto *= p_historial['alto'] if evidencia['historial'] == 'alto' else (1 - p_historial['alto'])
    score_bajo *= p_historial['bajo'] if evidencia['historial'] == 'alto' else (1 - p_historial['bajo'])

    total = score_alto + score_bajo
    return {'alto': score_alto / total, 'bajo': score_bajo / total}

proba = inferir_riesgo(evidencia)
proba
````

### 7.4 Decisión e interpretación

````python
prediccion = max(proba, key=proba.get)
print(f'Clase predicha: {prediccion}')
print(f'Probabilidad de riesgo alto: {proba['alto']:.3f}')
print(f'Probabilidad de riesgo bajo: {proba['bajo']:.3f}')
````

### 7.5 Resultados del modelo

Mostramos la inferencia probabilística obtenida para la evidencia definida.

````python
pd.DataFrame([proba])
````

### 7.6 Traducción de resultados

Interpretación en español de los resultados obtenidos:

- La red bayesiana asignó probabilidades posteriores a cada nivel de riesgo.
- La evidencia observada aumentó la probabilidad de la clase más compatible con los factores del caso.
- El modelo permite integrar conocimiento experto con evidencia parcial.
- Las probabilidades obtenidas son útiles para apoyar decisiones bajo incertidumbre.
- Al incorporar nueva evidencia, las creencias posteriores pueden cambiar de forma coherente.

### 7.7 Validación cuantitativa sobre un conjunto de prueba

Más allá de un único caso, evaluamos la red sobre un **conjunto de prueba sintético etiquetado** (cada caso lleva su nivel de riesgo real). Calculamos **exactitud**, **matriz de confusión** y **precision/recall/F1** — las métricas descritas en la sección 6. Los casos marcados como *difíciles* representan perfiles ambiguos en los que el modelo puede equivocarse.

````python
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

# Conjunto de prueba sintético: evidencia + nivel de riesgo real
casos_prueba = [
    {'ingresos': 'alto', 'deuda': 'alto', 'historial': 'alto', 'real': 'alto'},
    {'ingresos': 'bajo', 'deuda': 'bajo', 'historial': 'bajo', 'real': 'bajo'},
    {'ingresos': 'alto', 'deuda': 'bajo', 'historial': 'bajo', 'real': 'bajo'},
    {'ingresos': 'bajo', 'deuda': 'alto', 'historial': 'alto', 'real': 'alto'},
    {'ingresos': 'alto', 'deuda': 'alto', 'historial': 'bajo', 'real': 'alto'},  # caso difícil
    {'ingresos': 'bajo', 'deuda': 'bajo', 'historial': 'alto', 'real': 'alto'},
    {'ingresos': 'alto', 'deuda': 'bajo', 'historial': 'alto', 'real': 'bajo'},
    {'ingresos': 'bajo', 'deuda': 'alto', 'historial': 'bajo', 'real': 'bajo'},  # caso difícil
]

y_real, y_pred = [], []
for caso in casos_prueba:
    ev = {k: caso[k] for k in ('ingresos', 'deuda', 'historial')}
    p = inferir_riesgo(ev)
    y_real.append(caso['real'])
    y_pred.append(max(p, key=p.get))

acc = accuracy_score(y_real, y_pred)
print(f'Exactitud (accuracy) sobre {len(casos_prueba)} casos: {acc:.2%}\n')
print('Matriz de confusión (filas=real, columnas=predicho) — orden [bajo, alto]:')
print(confusion_matrix(y_real, y_pred, labels=['bajo', 'alto']))
print('\nReporte de clasificación:')
print(classification_report(y_real, y_pred, labels=['bajo', 'alto'], zero_division=0))
````

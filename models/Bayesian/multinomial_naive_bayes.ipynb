## 1. Descripción

**Multinomial Naive Bayes** es una variante del clasificador Naive Bayes diseñada para datos de conteo, especialmente texto representado mediante frecuencias de palabras.

Dada una instancia $X = (x_1, x_2, \\dots, x_m)$ donde cada atributo representa un conteo, el modelo estima la clase más probable usando probabilidades posteriores basadas en un modelo multinomial.

Es uno de los clasificadores más usados en tareas de texto por su simplicidad, velocidad y buen rendimiento en problemas de alta dimensionalidad dispersa.

### Casos de uso típicos
- Clasificación de spam.
- Clasificación de noticias o temas.
- Análisis de sentimiento con bolsa de palabras.
- Categorización de documentos.
- Filtrado de contenido textual.

## 4. Algoritmo de Entrenamiento

Multinomial Naive Bayes modela el conteo de atributos como una distribución multinomial condicionada a la clase.

### Pseudocódigo
```text
Entrada: D = {(x^(i), c^(i))}
Salida: clasificador multinomial

1. Contar la frecuencia de cada clase.
2. Para cada clase y cada atributo, contar ocurrencias.
3. Estimar probabilidades condicionales con suavizado.
4. Para un documento nuevo x, calcular:
   P(C | x) ∝ P(C) ∏ P(x_j | C)^{x_j}
5. Elegir la clase con mayor probabilidad posterior.
```

### Métricas clave

- **Probabilidad posterior:** $P(C \mid X)$
- **Exactitud:** porcentaje de aciertos
- **Precisión:** calidad de las predicciones positivas
- **Recall:** cobertura de los positivos reales
- **F1-score:** equilibrio entre precisión y recall

## 5. Supuestos y Restricciones

- **Datos de conteo:** cada atributo representa frecuencias no negativas.
- **Independencia condicional aproximada:** las palabras o features se tratan como independientes dado la clase.
- **Etiquetas disponibles:** se requiere un conjunto de entrenamiento etiquetado.
- **Texto preprocesado:** normalmente se elimina ruido, signos y stopwords.
- **Datos dispersos:** el modelo funciona muy bien en vectores de alta dimensión dispersa.
- **Suavizado recomendado:** evita probabilidades nulas en atributos no observados.

## 6. Tests / Métricas de validación

Multinomial Naive Bayes se valida mediante métricas de clasificación:

- **Exactitud (accuracy)** — porcentaje total de aciertos.
- **Precisión (precision)** — utilidad de las predicciones positivas.
- **Recall** — cobertura de las clases relevantes.
- **F1-score** — balance entre precisión y recall.
- **Matriz de confusión** — distribución de errores y aciertos.
- **Probabilidades posteriores** — interpretación de confianza por clase.

## 7. Implementación práctica

Usaremos un pequeño ejemplo textual para ilustrar Multinomial Naive Bayes.

### 7.1 Instalación e imports

````python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix

print('Librerías cargadas correctamente')
````

### 7.2 Dataset de ejemplo

Crearemos un pequeño conjunto de textos para clasificación binaria entre spam y no spam.

````python
textos = [
    'gana dinero rapido ahora',
    'oferta exclusiva compra ya',
    'reunion programada para manana',
    'tu factura fue enviada',
    'haz clic aqui para reclamar premio',
    'tu paquete ha sido entregado',
    'descuento especial solo hoy',
    'confirmacion de asistencia a la clase',
    'compra barata y envio gratis',
    'recordatorio de cita medica',
    'premio garantizado reclama ahora',
    'agenda la reunion con el equipo',
]

etiquetas = ['spam', 'spam', 'no_spam', 'no_spam', 'spam', 'no_spam', 'spam', 'no_spam', 'spam', 'no_spam', 'spam', 'no_spam']

df = pd.DataFrame({'texto': textos, 'clase': etiquetas})
df
````

### 7.3 Vectorización de texto

Convertimos los textos a conteos de palabras con bolsa de palabras.

````python
vectorizador = CountVectorizer()
X = vectorizador.fit_transform(df['texto'])
y = df['clase']

print('Vocabulario:', vectorizador.get_feature_names_out())
print('Matriz de conteos:', X.shape)
````

### 7.4 División entrenamiento y prueba

````python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.33, random_state=42, stratify=y)

print(f'Tamaño entrenamiento: {X_train.shape[0]}')
print(f'Tamaño prueba: {X_test.shape[0]}')
````

### 7.5 Entrenamiento del modelo

````python
modelo = MultinomialNB(alpha=1.0)
modelo.fit(X_train, y_train)
print('Modelo Multinomial Naive Bayes entrenado correctamente')
````

### 7.6 Predicción sobre el conjunto de prueba

````python
y_pred = modelo.predict(X_test)
resultado = pd.DataFrame({'real': y_test.values, 'prediccion': y_pred})
resultado['acierto'] = resultado['real'] == resultado['prediccion']
resultado
````

### 7.7 Métricas de evaluación

````python
accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred, pos_label='spam')
recall = recall_score(y_test, y_pred, pos_label='spam')
f1 = f1_score(y_test, y_pred, pos_label='spam')
cm = confusion_matrix(y_test, y_pred, labels=['spam', 'no_spam'])

print(f'Exactitud: {accuracy:.3f}')
print(f'Precisión: {precision:.3f}')
print(f'Recall: {recall:.3f}')
print(f'F1-score: {f1:.3f}')

pd.DataFrame(cm, index=['Real: spam', 'Real: no_spam'], columns=['Pred: spam', 'Pred: no_spam'])
````

### 7.8 Resultados del modelo

Mostramos las probabilidades posteriores estimadas para las muestras de prueba.

````python
probabilidades = modelo.predict_proba(X_test)
pd.DataFrame(probabilidades, columns=modelo.classes_)
````

### 7.9 Traducción de resultados

Interpretación en español de los resultados obtenidos:

- El modelo asigna probabilidades a cada clase a partir de los conteos de palabras.
- Las predicciones reflejan qué clase tiene mayor probabilidad posterior.
- La matriz de confusión muestra cuántos mensajes spam y no spam fueron identificados correctamente.
- La exactitud resume el desempeño global del clasificador.
- Las probabilidades ayudan a interpretar la fuerza de la decisión en cada documento.

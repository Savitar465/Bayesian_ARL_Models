## 1. Descripción

Una **Bayesian Belief Network (BBN)** es un modelo gráfico probabilístico que representa variables aleatorias y sus dependencias condicionales mediante un grafo dirigido acíclico. Cada nodo corresponde a una variable y cada arco codifica una relación de dependencia entre variables.

Las BBN permiten realizar inferencia probabilística, diagnóstico, predicción y análisis causal aproximado en dominios donde la incertidumbre es importante.

### Casos de uso típicos
- Diagnóstico médico.
- Sistemas expertos.
- Análisis de riesgo.
- Toma de decisiones bajo incertidumbre.
- Modelado causal en dominios tabulares.

## 3. Tipo de Modelo

| Criterio | Clasificación |
|---|---|
| **Método de aprendizaje** | Supervisado / aprendizaje probabilístico |
| **Por parámetros** | Paramétrico |
| **Datos de aprendizaje** | Offline (batch) |
| **Resultado del entrenamiento** | Red bayesiana con distribuciones condicionales |

Notas:
- **Supervisado** cuando se usa para inferir una variable objetivo a partir de evidencia observada.
- **Paramétrico** porque aprende un conjunto finito de tablas o distribuciones condicionales.
- **Offline** porque la estructura y parámetros se estiman desde una base de entrenamiento.
- El resultado es una estructura gráfica probabilística con capacidades de inferencia.

## 4. Algoritmo de Entrenamiento

El entrenamiento de una BBN incluye la definición de la estructura del grafo y la estimación de las probabilidades condicionales asociadas a cada nodo.

### Pseudocódigo
```text
Entrada: datos D, estructura G
Salida: red bayesiana parametrizada

1. Definir o aprender la estructura del grafo dirigido acíclico.
2. Para cada nodo X_i:
   a. Identificar sus nodos padres Pa(X_i).
   b. Estimar la distribución condicional P(X_i | Pa(X_i)).
3. Guardar las tablas de probabilidad condicional.
4. Para inferencia, propagar evidencia y calcular posteriors.
```

### Componentes principales
- **Nodos:** variables aleatorias.
- **Arcos dirigidos:** dependencias condicionales.
- **Tablas de probabilidad condicional (CPT):** parámetros del modelo.
- **Inferencia:** cálculo de probabilidades posteriores dadas evidencias.

### Métricas clave
- **Probabilidad posterior:** $P(H id E)$
- **Verosimilitud:** capacidad de explicar los datos observados
- **Exactitud diagnóstica:** proporción de aciertos en la variable objetivo
- **Log-loss:** calidad probabilística de las predicciones

## 5. Supuestos y Restricciones

- **Grafo acíclico dirigido:** no deben existir ciclos.
- **Variables bien definidas:** cada nodo representa una variable aleatoria.
- **Estructura correcta o aproximada:** la calidad depende de la red especificada o aprendida.
- **Datos suficientes:** las CPT necesitan suficientes ejemplos para ser estables.
- **Independencia condicional local:** cada nodo es independiente de sus no descendientes dado sus padres.
- **Complejidad computacional:** la inferencia exacta puede ser costosa en redes grandes.
- **Manejo de evidencia faltante:** los datos incompletos requieren tratamiento especial.

## 6. Tests / Métricas de validación

Las redes bayesianas se validan mediante métricas probabilísticas y de clasificación:

- **Exactitud** — aciertos sobre la variable objetivo.
- **Precision / Recall / F1** — en tareas de clasificación inferida.
- **Log-loss** — calidad de las probabilidades estimadas.
- **Matriz de confusión** — errores y aciertos por clase.
- **Consistencia de inferencia** — estabilidad de la posterior al incorporar evidencia.

## 7. Implementación práctica

Usaremos una red pequeña de ejemplo para realizar inferencia bayesiana manual.

### 7.1 Instalación e imports

````python
import pandas as pd

print('Librerías cargadas correctamente')
````

### 7.2 Red de ejemplo

Consideraremos una red simple para diagnosticar fiebre a partir de síntomas observados.

````python
# Probabilidades del modelo
prior_fiebre = {'si': 0.3, 'no': 0.7}

# CPTs: P(sintoma | fiebre)
p_tos = {'si': 0.8, 'no': 0.2}
p_dolor = {'si': 0.7, 'no': 0.1}
p_cansancio = {'si': 0.9, 'no': 0.3}

evidencia = {'tos': 'si', 'dolor': 'si', 'cansancio': 'si'}
evidencia
````

### 7.3 Inferencia bayesiana

````python
def posterior_fiebre(evidencia):
    score_si = prior_fiebre['si']
    score_no = prior_fiebre['no']

    for sintoma, valor in evidencia.items():
        if sintoma == 'tos':
            score_si *= p_tos['si'] if valor == 'si' else (1 - p_tos['si'])
            score_no *= p_tos['no'] if valor == 'si' else (1 - p_tos['no'])
        elif sintoma == 'dolor':
            score_si *= p_dolor['si'] if valor == 'si' else (1 - p_dolor['si'])
            score_no *= p_dolor['no'] if valor == 'si' else (1 - p_dolor['no'])
        elif sintoma == 'cansancio':
            score_si *= p_cansancio['si'] if valor == 'si' else (1 - p_cansancio['si'])
            score_no *= p_cansancio['no'] if valor == 'si' else (1 - p_cansancio['no'])

    total = score_si + score_no
    return {'si': score_si / total, 'no': score_no / total}

proba = posterior_fiebre(evidencia)
proba
````

### 7.4 Decisión e interpretación

````python
prediccion = max(proba, key=proba.get)
print(f'Clase predicha: {prediccion}')
print(f'Probabilidad de fiebre: {proba['si']:.3f}')
print(f'Probabilidad de no fiebre: {proba['no']:.3f}')
````

### 7.5 Resultados del modelo

Mostramos la inferencia probabilística obtenida para la evidencia definida.

````python
pd.DataFrame([proba])
````

### 7.6 Traducción de resultados

Interpretación en español de los resultados obtenidos:

- La evidencia observada incrementó la probabilidad de la clase más compatible con los síntomas.

- La salida probabilística es útil para diagnóstico y apoyo a decisiones.
- En una red bayesiana, cada nueva evidencia puede modificar las creencias posteriores.

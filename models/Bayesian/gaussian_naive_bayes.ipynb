## 1. Descripción

**Gaussian Naive Bayes** es una variante del clasificador Naive Bayes diseñada para atributos continuos. Asume que, dado cada clase, la distribución de cada atributo sigue una distribución normal.

Para una instancia $X = (x_1, x_2, \\dots, x_m)$, el modelo estima $P(C\mid X)$ a partir de medias y varianzas por clase y atributo. Es una opción natural cuando los datos son numéricos y la forma gaussiana es razonable.

### Casos de uso típicos
- Clasificación de flores o especies biológicas.
- Diagnóstico con variables numéricas.
- Reconocimiento de patrones en sensores.
- Modelos iniciales para datasets tabulares continuos.
- Clasificación rápida cuando se requiere un baseline interpretable.

## 3. Tipo de Modelo

| Criterio | Clasificación |
|---|---|
| **Método de aprendizaje** | Supervisado |
| **Por parámetros** | Paramétrico |
| **Datos de aprendizaje** | Offline (batch) |
| **Resultado del entrenamiento** | Clasificador probabilístico gaussiano |

Notas:
- **Supervisado** porque usa etiquetas de clase.
- **Paramétrico** porque estima medias y varianzas por clase.
- **Offline** porque aprende a partir de un conjunto de entrenamiento fijo.
- La salida es una distribución posterior por clase.

## 4. Algoritmo de Entrenamiento

Gaussian Naive Bayes estima la probabilidad de cada clase suponiendo que cada atributo continuo sigue una distribución normal condicionada a la clase.

### Pseudocódigo
```text
Entrada: D = {(x^(i), c^(i))}
Salida: clasificador gaussiano

1. Calcular la probabilidad previa P(C) para cada clase.
2. Para cada clase C y cada atributo j:
   - Calcular la media μ_{Cj}.
   - Calcular la varianza σ^2_{Cj}.
3. Para una nueva instancia x, calcular:
   P(C | x) ∝ P(C) ∏ N(x_j; μ_{Cj}, σ^2_{Cj})
4. Elegir la clase con mayor probabilidad posterior.
```

### Métricas clave

- **Probabilidad posterior:** $P(C \mid X)$
- **Exactitud:** proporción de predicciones correctas
- **Precisión:** calidad de las predicciones positivas
- **Recall:** cobertura de la clase positiva
- **F1-score:** equilibrio entre precisión y recall

## 5. Supuestos y Restricciones

- **Variables continuas** modeladas con distribución normal.
- **Independencia condicional aproximada** entre atributos.
- **Etiquetas de clase disponibles** durante el entrenamiento.
- **Sensibilidad a outliers:** valores extremos pueden afectar la media y la varianza.
- **Varianza no nula:** se requiere tratamiento cuando un atributo tiene poca variabilidad.
- **Atributos faltantes:** deben imputarse o tratarse antes del entrenamiento.

## 6. Tests / Métricas de validación

Gaussian Naive Bayes se evalúa como clasificador mediante métricas predictivas:

- **Exactitud (accuracy)** — porcentaje de aciertos.
- **Precisión (precision)** — calidad de las predicciones positivas.
- **Recall** — capacidad de detectar positivos reales.
- **F1-score** — media armónica entre precisión y recall.
- **Matriz de confusión** — resumen de errores por clase.
- **Probabilidades posteriores** — interpretación del nivel de confianza.

## 7. Implementación práctica

Usaremos el conjunto de datos Iris para ilustrar Gaussian Naive Bayes.

### 7.1 Instalación e imports

````python
import pandas as pd
import numpy as np
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix

print('Librerías cargadas correctamente')
````

### 7.2 Carga del dataset Iris

El conjunto Iris contiene variables numéricas y tres clases de flores.

````python
iris = load_iris(as_frame=True)
df = iris.frame.copy()
df['target_name'] = df['target'].map(dict(enumerate(iris.target_names)))
df.head()
````

### 7.3 División entrenamiento y prueba

````python
X = df[iris.feature_names]
y = df['target']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42, stratify=y)

print(f'Tamaño entrenamiento: {len(X_train)}')
print(f'Tamaño prueba: {len(X_test)}')
````

### 7.4 Entrenamiento del modelo

````python
modelo = GaussianNB()
modelo.fit(X_train, y_train)
print('Modelo Gaussian Naive Bayes entrenado correctamente')
````

### 7.5 Predicción sobre el conjunto de prueba

````python
y_pred = modelo.predict(X_test)
resultado = X_test.copy()
resultado['real'] = y_test.values
resultado['prediccion'] = y_pred
resultado['acierto'] = resultado['real'] == resultado['prediccion']
resultado.head()
````

### 7.6 Métricas de evaluación

````python
accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred, average='macro')
recall = recall_score(y_test, y_pred, average='macro')
f1 = f1_score(y_test, y_pred, average='macro')
cm = confusion_matrix(y_test, y_pred)

print(f'Exactitud: {accuracy:.3f}')
print(f'Precisión macro: {precision:.3f}')
print(f'Recall macro: {recall:.3f}')
print(f'F1-score macro: {f1:.3f}')

pd.DataFrame(cm, index=iris.target_names, columns=iris.target_names)
````

### 7.7 Resultados del modelo

Mostramos las probabilidades posteriores calculadas por el clasificador para algunas instancias de prueba.

````python
probabilidades = modelo.predict_proba(X_test[:5])
probabilidades_df = pd.DataFrame(probabilidades, columns=iris.target_names)
probabilidades_df
````

### 7.8 Traducción de resultados

Interpretación de los resultados obtenidos:

- Gaussian Naive Bayes modeló cada atributo numérico con una distribución normal por clase.
- Las probabilidades posteriores muestran qué clase considera más probable el modelo.
- La matriz de confusión permite observar en qué clases se producen más errores.
- Una exactitud alta indica que la distribución gaussiana es una buena aproximación para estos datos.
- Las métricas macro resumen el comportamiento global en las tres clases.

## 8. Conclusión

- Gaussian Naive Bayes es un clasificador **supervisado** adecuado para atributos continuos.
- Suponiendo normalidad por clase, estima medias y varianzas de forma eficiente.
- Es rápido, interpretable y útil como baseline para clasificación multiclase.
- Su desempeño depende de que la aproximación gaussiana sea razonable en los datos.
- En problemas tabulares numéricos, es una opción sólida y simple de implementar.
